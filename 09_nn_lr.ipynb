# Basic setting

### Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from keras import layers
from keras.models import Sequential
from sklearn.linear_model import LinearRegression

# Versions of TensorFlow and Keras
print(f"TensorFlow version: {tf.__version__}")
print(f"Keras version: {tf.keras.__version__}")
```

# Data collection

### Get data

```python
X_train = np.array([1, 2, 3, 4, 5, 6])
y_train = np.array([9, 12, 15, 18, 21, 24])
```

### Visualization

```python
plt.scatter(X_train, y_train)
plt.title("Training Data")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

### Reshape

```python
print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")

X_train = X_train.reshape(-1, 1)
print(f"Reshaped X_train: {X_train.shape}")
```

# Linear Regression

### Learning

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
print(f"Linear Regression Coef: {lr.coef_}, Intercept: {lr.intercept_}")
```

### Get prediction

```python
y_pred_s = lr.predict([[10]])
print(f"Linear Regression Prediction for X=10: {y_pred_s}")
```

# Neural Network

### Definition

```python
model = Sequential([
    layers.Input(shape=(1,)),  # Input layer
    layers.Dense(units=1)     # Dense layer
])
```

### Model structure

```python
model.summary()
```

### Compile

```python
model.compile(
    loss='mse',
    optimizer='adam',
    metrics=['mse']
)
```

### Learning

```python
history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=len(X_train),
    verbose=0
)
```

### weights from learning

```python
weights = model.get_weights()
print(f"Trained Weights: {weights}")
```

# Estimate

```python
y_pred = model.predict(np.array([[1], [5], [10]]))
print(f"Neural Network Predictions: {y_pred.flatten()}")
```

### Visualization

```python
plt.scatter(X_train, y_train, label="Data")
plt.plot(X_train, model.predict(X_train), 'r', label="Neural Network Fit")
plt.title("Model Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```
